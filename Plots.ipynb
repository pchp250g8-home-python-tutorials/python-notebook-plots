```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-10, 10, 0.1)
y = 1 / x
y[np.abs(x) < 0.01] = np.nan
plt.plot(x, y)
plt.ylim(-10, 10)

```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-10, 10, 0.1)
y = 7 ** (1 / (x - 5))
y[np.abs(x - 5) < 0.01]=np.nan
plt.plot(x, y)
plt.ylim(-10, 10)
```
