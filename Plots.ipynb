```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-10, 10, 0.1)
y = np.select([x != 5], [1 / x])
plt.plot(y.clip(-10, 10))

```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-10, 10, 0.1)
y = 7 ** (1 / (x - 5))
plt.plot(y.clip(-10, 10))
```
